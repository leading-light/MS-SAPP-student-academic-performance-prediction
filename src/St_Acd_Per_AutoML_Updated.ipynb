```python
# =============================================================================
# MODULE 1
# DEPENDENCIES AND LIBRARY IMPORTS
# =============================================================================

# Uncomment only if packages are not installed
# !pip install pandas numpy matplotlib seaborn scikit-learn
# !pip install catboost xgboost lightgbm openpyxl

# =============================================================================
# 1. GENERAL PURPOSE LIBRARIES
# =============================================================================

import warnings
warnings.filterwarnings("ignore")

import os
import time
import numpy as np
import pandas as pd

# =============================================================================
# 2. DATA VISUALIZATION
# =============================================================================

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
sns.set_theme(style="whitegrid")

# =============================================================================
# 3. DATA PREPROCESSING
# =============================================================================

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.preprocessing import (
    LabelEncoder,
    StandardScaler,
    label_binarize
)

from sklearn.impute import SimpleImputer

# =============================================================================
# 4. CLASSICAL MACHINE LEARNING CLASSIFIERS
# =============================================================================

from sklearn.linear_model import (
    LogisticRegression,
    PassiveAggressiveClassifier,
    SGDClassifier
)

from sklearn.tree import (
    DecisionTreeClassifier,
    ExtraTreeClassifier
)

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier,
    BaggingClassifier
)

from sklearn.neighbors import KNeighborsClassifier

from sklearn.svm import (
    SVC,
    LinearSVC
)

from sklearn.naive_bayes import (
    GaussianNB,
    BernoulliNB,
    MultinomialNB
)

from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis,
    QuadraticDiscriminantAnalysis
)

from sklearn.neural_network import MLPClassifier

# =============================================================================
# 5. ADVANCED BOOSTING MODELS
# =============================================================================

from catboost import CatBoostClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

# =============================================================================
# 6. PERFORMANCE EVALUATION METRICS
# =============================================================================

from sklearn.metrics import (

    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,

    matthews_corrcoef,
    cohen_kappa_score,

    roc_auc_score,
    roc_curve,
    auc,
    precision_recall_curve,

    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# =============================================================================
# 7. DISPLAY SETTINGS
# =============================================================================

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

# =============================================================================
# 8. RANDOM SEED
# =============================================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)

print("=" * 70)
print("Environment Ready")
print("=" * 70)

print("\nLibraries imported successfully.")
print("21 classifier dependencies are ready.")

# =============================================================================
# MODULE 2
# DATASET LOADING & INITIAL INSPECTION
# =============================================================================

#------------------------------------------------------------------------------
# Specify the dataset path
#------------------------------------------------------------------------------

file_path = r"E:\2026.02.03 D Backup From Previous SSD\Dropbox 15.10.22\after phd\Papers\Students academic performance prediction\Students academic performance prediction.csv"

#------------------------------------------------------------------------------
# Load dataset
#------------------------------------------------------------------------------

df = pd.read_csv(file_path)

print("="*80)
print("DATASET LOADED SUCCESSFULLY")
print("="*80)

#------------------------------------------------------------------------------
# Display dataset dimensions
#------------------------------------------------------------------------------

print(f"\nNumber of Records : {df.shape[0]}")
print(f"Number of Features: {df.shape[1]}")

#------------------------------------------------------------------------------
# Display first five records
#------------------------------------------------------------------------------

print("\nFirst Five Records")
display(df.head())

#------------------------------------------------------------------------------
# Display last five records
#------------------------------------------------------------------------------

print("\nLast Five Records")
display(df.tail())

#------------------------------------------------------------------------------
# Display column names
#------------------------------------------------------------------------------

print("\nDataset Columns")
print(df.columns.tolist())

#------------------------------------------------------------------------------
# Display data types
#------------------------------------------------------------------------------

print("\nFeature Data Types")
display(df.dtypes)

#------------------------------------------------------------------------------
# Dataset information
#------------------------------------------------------------------------------

print("\nDataset Information")
df.info()

#------------------------------------------------------------------------------
# Statistical summary (Numerical Features)
#------------------------------------------------------------------------------

print("\nStatistical Summary (Numerical Features)")
display(df.describe())

#------------------------------------------------------------------------------
# Statistical summary (Categorical Features)
#------------------------------------------------------------------------------

print("\nStatistical Summary (Categorical Features)")
display(df.describe(include="object"))

#------------------------------------------------------------------------------
# Check missing values
#------------------------------------------------------------------------------

print("\nMissing Values in Each Column")

missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage (%)": round(df.isnull().mean()*100,2)
})

display(missing)

#------------------------------------------------------------------------------
# Check duplicate rows
#------------------------------------------------------------------------------

duplicate_count = df.duplicated().sum()

print(f"\nDuplicate Records : {duplicate_count}")

#------------------------------------------------------------------------------
# Display target class distribution
#------------------------------------------------------------------------------

target_column = "CD_Div"

print("\nTarget Class Distribution")

display(df[target_column].value_counts())

print("\nTarget Class Percentage")

display(round(df[target_column].value_counts(normalize=True)*100,2))

#------------------------------------------------------------------------------
# Plot class distribution
#------------------------------------------------------------------------------

plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x=target_column,
    order=df[target_column].value_counts().index
)

plt.title("Class Distribution")
plt.xlabel("Academic Performance")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

print("="*80)
print("Module 2 Completed Successfully")
print("="*80)




# =============================================================================
# MODULE 3
# DATA CLEANING & DATASET VERSIONING
# PART 1 : ORIGINAL DATASET & DUPLICATE REMOVAL
# =============================================================================

print("=" * 80)
print("MODULE 3 (PART 1)")
print("DATASET VERSIONING")
print("=" * 80)

# =============================================================================
# STEP 1 : PRESERVE THE ORIGINAL DATASET
# =============================================================================

# Keep an untouched copy for future comparison

df_original = df.copy()

print("\nOriginal Dataset Created Successfully")

print(f"Rows    : {df_original.shape[0]}")
print(f"Columns : {df_original.shape[1]}")

# =============================================================================
# STEP 2 : IDENTIFY DUPLICATE RECORDS
# =============================================================================

duplicate_rows = df_original.duplicated()

duplicate_count = duplicate_rows.sum()

duplicate_percentage = duplicate_count / len(df_original) * 100

print("\nDuplicate Analysis")

print(f"Duplicate Records   : {duplicate_count}")

print(f"Duplicate Percentage: {duplicate_percentage:.2f}%")

# =============================================================================
# STEP 3 : DISPLAY DUPLICATE RECORDS (OPTIONAL)
# =============================================================================

if duplicate_count > 0:

    print("\nFirst Five Duplicate Records")

    display(df_original[duplicate_rows].head())

else:

    print("\nNo duplicate records were found.")

# =============================================================================
# STEP 4 : CREATE DUPLICATE-FREE DATASET
# =============================================================================

df_3class_duplicate_removed = (

    df_original
    .drop_duplicates()
    .reset_index(drop=True)

)

print("\nDuplicate-Free Dataset Created")

print(f"Rows    : {df_3class_duplicate_removed.shape[0]}")
print(f"Columns : {df_3class_duplicate_removed.shape[1]}")

# =============================================================================
# STEP 5 : DATASET COMPARISON
# =============================================================================

dataset_summary = pd.DataFrame({

    "Dataset":[

        "Original (3-Class)",

        "Duplicate Removed (3-Class)"

    ],

    "Rows":[

        df_original.shape[0],

        df_3class_duplicate_removed.shape[0]

    ],

    "Columns":[

        df_original.shape[1],

        df_3class_duplicate_removed.shape[1]

    ],

    "Duplicates Removed":[

        0,

        duplicate_count

    ]

})

print("\nDataset Summary")

display(dataset_summary)

# =============================================================================
# STEP 6 : CLASS DISTRIBUTION COMPARISON
# =============================================================================

target_column = "CD_Div"

print("\nOriginal Dataset Class Distribution")

display(df_original[target_column].value_counts())

print("\nDuplicate Removed Dataset Class Distribution")

display(df_3class_duplicate_removed[target_column].value_counts())

# =============================================================================
# MODULE COMPLETED
# =============================================================================

print("\nPart 1 Completed Successfully")


# =============================================================================
# MODULE 3
# DATA CLEANING & DATASET VERSIONING
# PART 2 : MISSING VALUE HANDLING & FEATURE ENCODING
# =============================================================================

from sklearn.preprocessing import LabelEncoder

print("=" * 80)
print("MODULE 3 (PART 2)")
print("MISSING VALUE HANDLING & FEATURE ENCODING")
print("=" * 80)

# =============================================================================
# STEP 7 : HANDLE MISSING VALUES
# =============================================================================

print("\nHandling Missing Values...")

#--------------------------------------------------------------------------
# Create copies so that previous datasets remain unchanged
#--------------------------------------------------------------------------

df_3class_original_encoded = df_original.copy()

df_3class_duplicate_encoded = df_3class_duplicate_removed.copy()

#--------------------------------------------------------------------------
# Function to impute missing values
#--------------------------------------------------------------------------

def impute_missing_values(dataframe):

    dataframe = dataframe.copy()

    for column in dataframe.columns:

        # Numerical Features
        if dataframe[column].dtype in ["int64", "float64"]:

            dataframe[column].fillna(
                dataframe[column].median(),
                inplace=True
            )

        # Categorical Features
        else:

            dataframe[column].fillna(
                dataframe[column].mode()[0],
                inplace=True
            )

    return dataframe

#--------------------------------------------------------------------------
# Apply Missing Value Imputation
#--------------------------------------------------------------------------

df_3class_original_encoded = impute_missing_values(
    df_3class_original_encoded
)

df_3class_duplicate_encoded = impute_missing_values(
    df_3class_duplicate_encoded
)

print("Missing values handled successfully.")

#--------------------------------------------------------------------------
# Verify Missing Values
#--------------------------------------------------------------------------

print("\nRemaining Missing Values (Original Dataset)")

display(df_3class_original_encoded.isnull().sum())

print("\nRemaining Missing Values (Duplicate Removed Dataset)")

display(df_3class_duplicate_encoded.isnull().sum())

# =============================================================================
# STEP 8 : ENCODE CATEGORICAL FEATURES
# =============================================================================

print("\nEncoding Categorical Features...")

target_column = "CD_Div"

#--------------------------------------------------------------------------
# Feature Encoders
#--------------------------------------------------------------------------

feature_encoders_original = {}

feature_encoders_duplicate = {}

#--------------------------------------------------------------------------
# Encode Original Dataset
#--------------------------------------------------------------------------

X_original_encoded = df_3class_original_encoded.drop(
    columns=[target_column]
).copy()

y_original_encoded = df_3class_original_encoded[target_column].copy()

for column in X_original_encoded.columns:

    if X_original_encoded[column].dtype == "object":

        encoder = LabelEncoder()

        X_original_encoded[column] = encoder.fit_transform(
            X_original_encoded[column].astype(str)
        )

        feature_encoders_original[column] = encoder

#--------------------------------------------------------------------------
# Encode Duplicate Removed Dataset
#--------------------------------------------------------------------------

X_duplicate_encoded = df_3class_duplicate_encoded.drop(
    columns=[target_column]
).copy()

y_duplicate_encoded = df_3class_duplicate_encoded[target_column].copy()

for column in X_duplicate_encoded.columns:

    if X_duplicate_encoded[column].dtype == "object":

        encoder = LabelEncoder()

        X_duplicate_encoded[column] = encoder.fit_transform(
            X_duplicate_encoded[column].astype(str)
        )

        feature_encoders_duplicate[column] = encoder

#--------------------------------------------------------------------------
# Encode Target Variable
#--------------------------------------------------------------------------

target_encoder = LabelEncoder()

y_original_encoded = target_encoder.fit_transform(
    y_original_encoded
)

y_duplicate_encoded = target_encoder.transform(
    y_duplicate_encoded
)

print("Categorical Encoding Completed.")

# =============================================================================
# STEP 9 : DISPLAY DATASET INFORMATION
# =============================================================================

print("\nEncoded Original Dataset")

print(X_original_encoded.shape)

print("\nEncoded Duplicate Removed Dataset")

print(X_duplicate_encoded.shape)

print("\nTarget Classes")

print(target_encoder.classes_)

# =============================================================================
# STEP 10 : DISPLAY FEATURE TYPES
# =============================================================================

feature_summary = pd.DataFrame({

    "Feature": X_duplicate_encoded.columns,

    "Data Type": X_duplicate_encoded.dtypes.values

})

print("\nEncoded Feature Summary")

display(feature_summary)

# =============================================================================
# STEP 11 : PREPROCESSING SUMMARY
# =============================================================================

summary = pd.DataFrame({

    "Dataset":[

        "Original (Encoded)",

        "Duplicate Removed (Encoded)"

    ],

    "Rows":[

        X_original_encoded.shape[0],

        X_duplicate_encoded.shape[0]

    ],

    "Features":[

        X_original_encoded.shape[1],

        X_duplicate_encoded.shape[1]

    ]

})

print("\nDataset Summary")

display(summary)

print("\nPart 2 Completed Successfully")


# =============================================================================
# MODULE 3
# DATA CLEANING & DATASET VERSIONING
# PART 3 : OUTLIER DETECTION & REMOVAL
# =============================================================================

from sklearn.ensemble import IsolationForest

print("="*80)
print("MODULE 3 (PART 3)")
print("OUTLIER DETECTION USING ISOLATION FOREST")
print("="*80)

# =============================================================================
# STEP 12 : PRESERVE DATASET BEFORE OUTLIER REMOVAL
# =============================================================================

X_before_outlier = X_duplicate_encoded.copy()
y_before_outlier = y_duplicate_encoded.copy()

print("\nDataset Before Outlier Removal")

print(f"Samples  : {X_before_outlier.shape[0]}")
print(f"Features : {X_before_outlier.shape[1]}")

# =============================================================================
# STEP 13 : BUILD ISOLATION FOREST MODEL
# =============================================================================

print("\nTraining Isolation Forest...")

iso_forest = IsolationForest(

    n_estimators=300,
    contamination="auto",
    random_state=42,
    n_jobs=-1

)

outlier_labels = iso_forest.fit_predict(X_before_outlier)

print("Outlier detection completed.")

# =============================================================================
# STEP 14 : IDENTIFY OUTLIERS
# =============================================================================

normal_count = np.sum(outlier_labels == 1)

outlier_count = np.sum(outlier_labels == -1)

print("\nOutlier Summary")

print(f"Normal Samples  : {normal_count}")

print(f"Outlier Samples : {outlier_count}")

print(f"Percentage      : {(outlier_count/len(outlier_labels))*100:.2f}%")

# =============================================================================
# STEP 15 : CREATE OUTLIER-FREE DATASET
# =============================================================================

X_outlier_removed = X_before_outlier[outlier_labels == 1].reset_index(drop=True)

y_outlier_removed = y_before_outlier[outlier_labels == 1]

print("\nDataset After Outlier Removal")

print(f"Samples  : {X_outlier_removed.shape[0]}")

print(f"Features : {X_outlier_removed.shape[1]}")

# =============================================================================
# STEP 16 : CREATE DATAFRAME VERSION
# =============================================================================

df_3class_outlier_removed = pd.concat(

    [

        X_outlier_removed,

        pd.Series(
            target_encoder.inverse_transform(y_outlier_removed),
            name=target_column
        )

    ],

    axis=1

)

print("\n3-Class Outlier-Free Dataset Created Successfully")

# =============================================================================
# STEP 17 : VISUALIZE OUTLIER DISTRIBUTION
# =============================================================================

plt.figure(figsize=(6,4))

sns.countplot(

    x=pd.Series(outlier_labels).map(

        {

            1:"Normal",

            -1:"Outlier"

        }

    )

)

plt.title("Isolation Forest Outlier Detection")

plt.xlabel("Observation")

plt.ylabel("Frequency")

plt.tight_layout()

plt.show()

# =============================================================================
# STEP 18 : CLASS DISTRIBUTION COMPARISON
# =============================================================================

print("\nClass Distribution BEFORE Outlier Removal")

display(

    pd.Series(

        target_encoder.inverse_transform(y_before_outlier)

    ).value_counts()

)

print("\nClass Distribution AFTER Outlier Removal")

display(

    pd.Series(

        target_encoder.inverse_transform(y_outlier_removed)

    ).value_counts()

)

# =============================================================================
# STEP 19 : DATASET SUMMARY
# =============================================================================

summary = pd.DataFrame({

    "Dataset":[

        "Duplicate Removed",

        "Outlier Removed"

    ],

    "Samples":[

        len(X_before_outlier),

        len(X_outlier_removed)

    ],

    "Features":[

        X_before_outlier.shape[1],

        X_outlier_removed.shape[1]

    ]

})

print("\nDataset Summary")

display(summary)

print("="*80)
print("PART 3 COMPLETED SUCCESSFULLY")
print("="*80)

# =============================================================================
# MODULE 3
# DATA CLEANING & DATASET VERSIONING
# PART 4 : CREATE BINARY (PASS/FAIL) DATASETS
# =============================================================================

print("=" * 80)
print("MODULE 3 (PART 4)")
print("CREATING BINARY (PASS / FAIL) DATASETS")
print("=" * 80)

# =============================================================================
# STEP 20 : CREATE 2-CLASS DATASET FROM ORIGINAL DATA
# =============================================================================

print("\nCreating 2-Class Dataset (Original)...")

df_2class_original = df_original.copy()

df_2class_original[target_column] = df_2class_original[target_column].replace({

    "A": "Pass",
    "B": "Pass",
    "F": "Fail"

})

# =============================================================================
# STEP 21 : CREATE 2-CLASS DATASET FROM DUPLICATE REMOVED DATA
# =============================================================================

print("Creating 2-Class Dataset (Duplicate Removed)...")

df_2class_duplicate_removed = df_3class_duplicate_removed.copy()

df_2class_duplicate_removed[target_column] = \
df_2class_duplicate_removed[target_column].replace({

    "A": "Pass",
    "B": "Pass",
    "F": "Fail"

})

# =============================================================================
# STEP 22 : CREATE 2-CLASS DATASET FROM OUTLIER REMOVED DATA
# =============================================================================

print("Creating 2-Class Dataset (Outlier Removed)...")

df_2class_outlier_removed = df_3class_outlier_removed.copy()

df_2class_outlier_removed[target_column] = \
df_2class_outlier_removed[target_column].replace({

    "A": "Pass",
    "B": "Pass",
    "F": "Fail"

})

# =============================================================================
# STEP 23 : DISPLAY CLASS DISTRIBUTIONS
# =============================================================================

print("\nOriginal Binary Dataset")

display(df_2class_original[target_column].value_counts())

print("\nDuplicate Removed Binary Dataset")

display(df_2class_duplicate_removed[target_column].value_counts())

print("\nOutlier Removed Binary Dataset")

display(df_2class_outlier_removed[target_column].value_counts())

# =============================================================================
# STEP 24 : DATASET INFORMATION
# =============================================================================

dataset_information = pd.DataFrame({

    "Dataset":[

        "Original (3-Class)",

        "Duplicate Removed (3-Class)",

        "Outlier Removed (3-Class)",

        "Original (2-Class)",

        "Duplicate Removed (2-Class)",

        "Outlier Removed (2-Class)"

    ],

    "Samples":[

        len(df_original),

        len(df_3class_duplicate_removed),

        len(df_3class_outlier_removed),

        len(df_2class_original),

        len(df_2class_duplicate_removed),

        len(df_2class_outlier_removed)

    ],

    "Classes":[

        3,

        3,

        3,

        2,

        2,

        2

    ]

})

print("\nDataset Summary")

display(dataset_information)

# =============================================================================
# STEP 25 : VISUALIZE ALL DATASETS
# =============================================================================

fig, axes = plt.subplots(2, 3, figsize=(16,8))

datasets = [

    df_original,

    df_3class_duplicate_removed,

    df_3class_outlier_removed,

    df_2class_original,

    df_2class_duplicate_removed,

    df_2class_outlier_removed

]

titles = [

    "Original (3-Class)",

    "Duplicate Removed",

    "Outlier Removed",

    "Original (2-Class)",

    "Duplicate Removed",

    "Outlier Removed"

]

for ax, data, title in zip(axes.ravel(), datasets, titles):

    sns.countplot(

        x=target_column,

        data=data,

        ax=ax

    )

    ax.set_title(title)

    ax.set_xlabel("Class")

    ax.set_ylabel("Count")

plt.tight_layout()

plt.show()

# =============================================================================
# STEP 26 : STORE DATASETS IN A DICTIONARY
# =============================================================================

datasets = {

    #------------------------------------------------------------
    # Three-Class Datasets
    #------------------------------------------------------------

    "3class_original": df_original,

    "3class_duplicate_removed": df_3class_duplicate_removed,

    "3class_outlier_removed": df_3class_outlier_removed,

    #------------------------------------------------------------
    # Two-Class Datasets
    #------------------------------------------------------------

    "2class_original": df_2class_original,

    "2class_duplicate_removed": df_2class_duplicate_removed,

    "2class_outlier_removed": df_2class_outlier_removed

}

print("\nDatasets Dictionary Created Successfully.")

print("\nAvailable Datasets")

for key in datasets.keys():

    print("•", key)

# =============================================================================
# MODULE 3 COMPLETED
# =============================================================================

print("\n" + "="*80)
print("MODULE 3 COMPLETED SUCCESSFULLY")
print("="*80)

print("\nSix datasets are now available for experimentation.")

# =============================================================================
# MODULE 4
# DATA PREPARATION FOR MACHINE LEARNING
# =============================================================================

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

print("=" * 80)
print("MODULE 4")
print("DATA PREPARATION FOR MACHINE LEARNING")
print("=" * 80)

# =============================================================================
# STEP 1 : DEFINE ALL DATASETS
# =============================================================================

dataset_registry = {

    "3class_original": df_original,

    "3class_duplicate_removed": df_3class_duplicate_removed,

    "3class_outlier_removed": df_3class_outlier_removed,

    "2class_original": df_2class_original,

    "2class_duplicate_removed": df_2class_duplicate_removed,

    "2class_outlier_removed": df_2class_outlier_removed

}

print("\nDatasets Registered Successfully")

# =============================================================================
# STEP 2 : PREPARE EACH DATASET
# =============================================================================

prepared_datasets = {}

for dataset_name, dataframe in dataset_registry.items():

    print(f"\nPreparing : {dataset_name}")

    #-----------------------------------------------------------------------
    # Separate Features & Target
    #-----------------------------------------------------------------------

    X = dataframe.drop(columns=[target_column]).copy()

    y = dataframe[target_column].copy()

    #-----------------------------------------------------------------------
    # Encode categorical features
    #-----------------------------------------------------------------------

    feature_encoders = {}

    for column in X.columns:

        if X[column].dtype == "object":

            encoder = LabelEncoder()

            X[column] = encoder.fit_transform(
                X[column].astype(str)
            )

            feature_encoders[column] = encoder

    #-----------------------------------------------------------------------
    # Encode target labels
    #-----------------------------------------------------------------------

    target_encoder = LabelEncoder()

    y_encoded = target_encoder.fit_transform(y)

    #-----------------------------------------------------------------------
    # Stratified Train/Test Split
    #-----------------------------------------------------------------------

    X_train, X_test, y_train, y_test = train_test_split(

        X,

        y_encoded,

        test_size=0.20,

        random_state=42,

        stratify=y_encoded

    )

    #-----------------------------------------------------------------------
    # Store everything
    #-----------------------------------------------------------------------

    prepared_datasets[dataset_name] = {

        "X_train": X_train,

        "X_test": X_test,

        "y_train": y_train,

        "y_test": y_test,

        "feature_names": list(X.columns),

        "feature_encoders": feature_encoders,

        "target_encoder": target_encoder,

        "X": X,

        "y": y_encoded

    }

print("\nAll datasets prepared successfully.")

# =============================================================================
# STEP 3 : DISPLAY SUMMARY
# =============================================================================

summary = []

for name in prepared_datasets.keys():

    obj = prepared_datasets[name]

    summary.append({

        "Dataset": name,

        "Samples": obj["X"].shape[0],

        "Features": obj["X"].shape[1],

        "Training": obj["X_train"].shape[0],

        "Testing": obj["X_test"].shape[0],

        "Classes": len(obj["target_encoder"].classes_)

    })

summary = pd.DataFrame(summary)

display(summary)

print("\nMODULE 4 COMPLETED SUCCESSFULLY")


# =============================================================================
# MODULE 5
# MODEL EVALUATION FRAMEWORK
# =============================================================================

print("=" * 80)
print("MODULE 5 : MODEL EVALUATION FRAMEWORK")
print("=" * 80)

overall_results = []
classwise_results = []

print("Evaluation framework initialized.")


# =============================================================================
# FUNCTION 1 : OBTAIN SCORE / PROBABILITY OUTPUT
# =============================================================================

def get_model_scores(model, X_test):

    # ---------------------------------------------------------
    # Case 1 : Probability-based classifiers
    # ---------------------------------------------------------

    if hasattr(model, "predict_proba"):

        y_score = model.predict_proba(X_test)

        return y_score, "Probability"

    # ---------------------------------------------------------
    # Case 2 : Decision-function classifiers
    # LinearSVC / PassiveAggressive / SGD
    # ---------------------------------------------------------

    elif hasattr(model, "decision_function"):

        decision_scores = model.decision_function(X_test)

        # Binary classification
        if decision_scores.ndim == 1:

            return decision_scores, "Decision Score"

        # Multiclass classification
        else:

            return decision_scores, "Decision Score"

    else:

        return None, "No Score"


# =============================================================================
# FUNCTION 2 : EVALUATE MODEL
# =============================================================================

def evaluate_model(
        model,
        model_name,
        dataset_name,
        X_test,
        y_test,
        target_encoder,
        train_time):

    # ---------------------------------------------------------
    # Prediction
    # ---------------------------------------------------------

    start = time.time()

    y_pred = model.predict(X_test)

    prediction_time = time.time() - start

    # ---------------------------------------------------------
    # Obtain probability / decision scores
    # ---------------------------------------------------------

    y_score, score_type = get_model_scores(
        model,
        X_test
    )

    class_names = target_encoder.classes_

    # ---------------------------------------------------------
    # Overall Metrics
    # ---------------------------------------------------------

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    balanced_acc = balanced_accuracy_score(
        y_test,
        y_pred
    )

    precision_macro = precision_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    precision_weighted = precision_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    recall_macro = recall_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall_weighted = recall_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    f1_macro = f1_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1_weighted = f1_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    mcc = matthews_corrcoef(
        y_test,
        y_pred
    )

    kappa = cohen_kappa_score(
        y_test,
        y_pred
    )

    # ---------------------------------------------------------
    # ROC AUC
    # ---------------------------------------------------------

    roc_auc = np.nan

    try:

        if y_score is not None:

            # Binary classification
            if len(class_names) == 2:

                if y_score.ndim == 1:

                    roc_auc = roc_auc_score(
                        y_test,
                        y_score
                    )

                else:

                    roc_auc = roc_auc_score(
                        y_test,
                        y_score[:, 1]
                    )

            # Multiclass classification
            else:

                y_bin = label_binarize(
                    y_test,
                    classes=np.arange(len(class_names))
                )

                roc_auc = roc_auc_score(
                    y_bin,
                    y_score,
                    average="weighted",
                    multi_class="ovr"
                )

    except Exception as e:

        print(
            f"ROC AUC could not be calculated for "
            f"{model_name}: {e}"
        )

        roc_auc = np.nan

    # ---------------------------------------------------------
    # Save Overall Results
    # ---------------------------------------------------------

    overall_results.append({

        "Dataset": dataset_name,

        "Model": model_name,

        "Accuracy": accuracy,

        "Balanced Accuracy": balanced_acc,

        "Precision (Macro)": precision_macro,

        "Precision (Weighted)": precision_weighted,

        "Recall (Macro)": recall_macro,

        "Recall (Weighted)": recall_weighted,

        "F1 (Macro)": f1_macro,

        "F1 (Weighted)": f1_weighted,

        "MCC": mcc,

        "Kappa": kappa,

        "ROC AUC": roc_auc,

        "Score Type": score_type,

        "Training Time": train_time,

        "Prediction Time": prediction_time

    })

    # ---------------------------------------------------------
    # Class-wise Results
    # ---------------------------------------------------------

    cm = confusion_matrix(
        y_test,
        y_pred
    )

    for i, cls in enumerate(class_names):

        TP = cm[i, i]

        FP = cm[:, i].sum() - TP

        FN = cm[i, :].sum() - TP

        TN = cm.sum() - TP - FP - FN

        sensitivity = (
            TP / (TP + FN)
            if (TP + FN)
            else 0
        )

        specificity = (
            TN / (TN + FP)
            if (TN + FP)
            else 0
        )

        precision = (
            TP / (TP + FP)
            if (TP + FP)
            else 0
        )

        f1 = (

            2 * precision * sensitivity /
            (precision + sensitivity)

            if (precision + sensitivity)

            else 0

        )

        classwise_results.append({

            "Dataset": dataset_name,

            "Model": model_name,

            "Class": cls,

            "TP": TP,

            "TN": TN,

            "FP": FP,

            "FN": FN,

            "Precision": precision,

            "Sensitivity": sensitivity,

            "Specificity": specificity,

            "F1": f1

        })

    # ---------------------------------------------------------
    # Print Summary
    # ---------------------------------------------------------

    print("\n")
    print("=" * 70)
    print(model_name)
    print("=" * 70)

    print(

        classification_report(
            y_test,
            y_pred,
            target_names=class_names,
            zero_division=0
        )

    )

    ConfusionMatrixDisplay.from_predictions(

        y_test,
        y_pred,

        display_labels=class_names,

        cmap="Blues"

    )

    plt.title(
        f"{model_name} - {dataset_name}"
    )

    plt.show()


print("MODULE 5 COMPLETED SUCCESSFULLY")



# =============================================================================
# MODULE 6
# DEFINE BASELINE CLASSIFIERS
# =============================================================================

print("=" * 80)
print("MODULE 6 : DEFINE BASELINE CLASSIFIERS")
print("=" * 80)

baseline_models = {

    # -------------------------------------------------------------------------
    # Linear Models
    # -------------------------------------------------------------------------

    "Logistic Regression":
        LogisticRegression(
            max_iter=1000,
            random_state=42
        ),

    "Passive Aggressive":
        PassiveAggressiveClassifier(
            max_iter=1000,
            random_state=42
        ),

    "Stochastic Gradient Descent":
        SGDClassifier(
            max_iter=1000,
            random_state=42
        ),

    # -------------------------------------------------------------------------
    # Tree Models
    # -------------------------------------------------------------------------

    "Decision Tree":
        DecisionTreeClassifier(
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            n_jobs=-1
        ),

    "Extra Trees":
        ExtraTreesClassifier(
            n_estimators=300,
            random_state=42,
            n_jobs=-1
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            random_state=42
        ),

    "AdaBoost":
        AdaBoostClassifier(
            random_state=42
        ),

    "Bagging":
        BaggingClassifier(
            random_state=42
        ),

    # -------------------------------------------------------------------------
    # Nearest Neighbour
    # -------------------------------------------------------------------------

    "K-Nearest Neighbors":
        KNeighborsClassifier(
            n_neighbors=5
        ),

    # -------------------------------------------------------------------------
    # Naive Bayes
    # -------------------------------------------------------------------------

    "Gaussian Naive Bayes":
        GaussianNB(),

    "Bernoulli Naive Bayes":
        BernoulliNB(),

    "Multinomial Naive Bayes":
        MultinomialNB(),

    # -------------------------------------------------------------------------
    # Discriminant Analysis
    # -------------------------------------------------------------------------

    "Linear Discriminant Analysis":
        LinearDiscriminantAnalysis(),

    "Quadratic Discriminant Analysis":
        QuadraticDiscriminantAnalysis(),

    # -------------------------------------------------------------------------
    # Support Vector Machines
    # -------------------------------------------------------------------------

    "Support Vector Machine":
        SVC(
            probability=True,
            random_state=42
        ),

    "Linear SVM":
        LinearSVC(
            max_iter=5000,
            random_state=42
        ),

    # -------------------------------------------------------------------------
    # Neural Network
    # -------------------------------------------------------------------------

    "Multi-layer Perceptron":
        MLPClassifier(
            hidden_layer_sizes=(100,),
            max_iter=500,
            random_state=42
        ),

    # -------------------------------------------------------------------------
    # Advanced Boosting
    # -------------------------------------------------------------------------

    "CatBoost":
        CatBoostClassifier(
            verbose=0,
            random_state=42
        ),

    "XGBoost":
        XGBClassifier(
            random_state=42,
            eval_metric="mlogloss"
        ),

    "LightGBM":
        LGBMClassifier(
            random_state=42,
            verbose=-1
        )

}

print(
    f"\nTotal classifiers defined : "
    f"{len(baseline_models)}"
)

print("\nClassifiers:")

for i, model_name in enumerate(
    baseline_models.keys(),
    start=1
):

    print(f"{i:02d}. {model_name}")

print("\nExpected experiments:")
print(
    f"{len(prepared_datasets)} datasets × "
    f"{len(baseline_models)} classifiers = "
    f"{len(prepared_datasets) * len(baseline_models)} experiments"
)

print("\nMODULE 6 COMPLETED SUCCESSFULLY")


# =============================================================================
# MODULE 7
# TRAIN & EVALUATE ALL BASELINE CLASSIFIERS
# =============================================================================

import time

print("=" * 80)
print("MODULE 7 : TRAINING & EVALUATION")
print("=" * 80)

experiment_counter = 1
total_experiments = len(prepared_datasets) * len(baseline_models)

for dataset_name, dataset in prepared_datasets.items():

    print("\n" + "="*80)
    print(f"Dataset : {dataset_name}")
    print("="*80)

    X_train = dataset["X_train"]
    X_test  = dataset["X_test"]

    y_train = dataset["y_train"]
    y_test  = dataset["y_test"]

    target_encoder = dataset["target_encoder"]

    #----------------------------------------------------------
    # Train every classifier
    #----------------------------------------------------------

    for model_name, model in baseline_models.items():

        print(f"\n[{experiment_counter}/{total_experiments}] {model_name}")

        try:

            #--------------------------------------------------
            # Training
            #--------------------------------------------------

            start_train = time.time()

            model.fit(X_train, y_train)

            train_time = time.time() - start_train

            #--------------------------------------------------
            # Evaluation
            #--------------------------------------------------

            evaluate_model(

                model=model,

                model_name=model_name,

                dataset_name=dataset_name,

                X_test=X_test,

                y_test=y_test,

                target_encoder=target_encoder,

                train_time=train_time

            )

        except Exception as e:

            print(f"Error in {model_name}")

            print(e)

        experiment_counter += 1

print("\n")
print("="*80)
print("="*80)

overall_results_df = pd.DataFrame(overall_results)

classwise_results_df = pd.DataFrame(classwise_results)

overall_results_df.to_csv(
    "Overall_Results.csv",
    index=False
)

classwise_results_df.to_csv(
    "Classwise_Results.csv",
    index=False
)
print("MODULE 7 COMPLETED SUCCESSFULLY")


# =============================================================================
# MODULE 8
# RESULTS ANALYSIS, COMPOSITE SCORE, RANKING,
# TOP-5 SELECTION, OVERLAP ANALYSIS & FINAL WINNER
# =============================================================================

print("=" * 90)
print("MODULE 8 : RESULTS ANALYSIS & FINAL MODEL SELECTION")
print("=" * 90)


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import numpy as np
import pandas as pd

from datetime import datetime


# =============================================================================
# 2. CREATE TIMESTAMPED RESULTS FOLDER
# =============================================================================

# IMPORTANT:
# Change "file_path" below ONLY if your Module 1 uses another variable name.

dataset_directory = os.path.dirname(file_path)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

results_folder = os.path.join(
    dataset_directory,
    "Results",
    timestamp
)

os.makedirs(
    results_folder,
    exist_ok=True
)

print("\nResults folder:")
print(results_folder)


# =============================================================================
# 3. CREATE RESULT DATAFRAMES
# =============================================================================

overall_results_df = pd.DataFrame(
    overall_results
)

classwise_results_df = pd.DataFrame(
    classwise_results
)


print("\nOverall experiments:", len(overall_results_df))
print("Classwise records:", len(classwise_results_df))


# =============================================================================
# 4. CLEAN NUMERIC COLUMNS
# =============================================================================

metric_columns = [

    "Accuracy",
    "Balanced Accuracy",

    "Precision (Macro)",
    "Precision (Weighted)",

    "Recall (Macro)",
    "Recall (Weighted)",

    "F1 (Macro)",
    "F1 (Weighted)",

    "MCC",
    "Kappa",

    "ROC AUC",

    "Training Time",
    "Prediction Time"

]


for column in metric_columns:

    if column in overall_results_df.columns:

        overall_results_df[column] = pd.to_numeric(
            overall_results_df[column],
            errors="coerce"
        )


# Replace infinite values

overall_results_df = (
    overall_results_df
    .replace([np.inf, -np.inf], np.nan)
)


# Ensure Model and Dataset are strings

overall_results_df["Model"] = (
    overall_results_df["Model"]
    .astype(str)
    .str.strip()
)

overall_results_df["Dataset"] = (
    overall_results_df["Dataset"]
    .astype(str)
    .str.strip()
)


# =============================================================================
# 5. NORMALIZED MCC
# =============================================================================
#
# MCC ranges from -1 to +1.
#
# Normalized MCC:
#
#       (MCC + 1) / 2
#
# Result ranges from 0 to 1.
# =============================================================================

overall_results_df["Normalized MCC"] = (

    (
        overall_results_df["MCC"] + 1
    ) / 2

)


# =============================================================================
# 6. COMPOSITE SCORE
# =============================================================================
#
# EXACT FORMULA:
#
# Composite Score =
#
# 0.30(Balanced Accuracy)
# + 0.30(F1 Weighted)
# + 0.20((MCC + 1)/2)
# + 0.20(ROC-AUC)
#
# =============================================================================

overall_results_df["Composite Score"] = (

      0.30
      * overall_results_df["Balanced Accuracy"]

    + 0.30
      * overall_results_df["F1 (Weighted)"]

    + 0.20
      * overall_results_df["Normalized MCC"]

    + 0.20
      * overall_results_df["ROC AUC"]

)


# Replace infinite values

overall_results_df["Composite Score"] = (

    overall_results_df["Composite Score"]
    .replace([np.inf, -np.inf], np.nan)

)


# Round

overall_results_df["Composite Score"] = (

    overall_results_df["Composite Score"]
    .round(4)

)


# =============================================================================
# 7. VALID / INVALID RESULTS
# =============================================================================

overall_results_df["Valid Composite"] = (

    overall_results_df["Composite Score"]
    .notna()

)


print("\nValid experiments:",
      overall_results_df["Valid Composite"].sum())

print("Invalid experiments:",
      (~overall_results_df["Valid Composite"]).sum())


# =============================================================================
# 8. DATASET-SPECIFIC RANKING
# =============================================================================

overall_results_df["Dataset Rank"] = (

    overall_results_df
    .groupby("Dataset")["Composite Score"]
    .rank(
        ascending=False,
        method="min",
        na_option="keep"
    )

)


# Nullable integer prevents NaN conversion errors

overall_results_df["Dataset Rank"] = (

    overall_results_df["Dataset Rank"]
    .astype("Int64")

)


# =============================================================================
# 9. COMPLETE RESULTS
# =============================================================================

all_results_df = (

    overall_results_df

    .sort_values(
        by=[
            "Dataset",
            "Composite Score"
        ],
        ascending=[
            True,
            False
        ],
        na_position="last"
    )

    .reset_index(drop=True)

)


print("\n")
print("=" * 90)
print("COMPLETE EXPERIMENT RESULTS")
print("=" * 90)

display(all_results_df)


# =============================================================================
# 10. VALID RESULTS ONLY FOR RANKING
# =============================================================================

valid_results_df = (

    overall_results_df[
        overall_results_df["Valid Composite"]
    ]

    .copy()

)


# =============================================================================
# 11. TOP-5 MODELS FOR EACH DATASET
# =============================================================================

top5_per_dataset = (

    valid_results_df

    .sort_values(
        by=[
            "Dataset",
            "Composite Score"
        ],
        ascending=[
            True,
            False
        ]
    )

    .groupby(
        "Dataset",
        group_keys=False
    )

    .head(5)

    .reset_index(drop=True)

)


top5_per_dataset["Top-5 Position"] = (

    top5_per_dataset
    .groupby("Dataset")
    .cumcount()
    + 1

)


print("\n")
print("=" * 90)
print("TOP-5 MODELS FOR EACH DATASET")
print("=" * 90)

display(top5_per_dataset)


# =============================================================================
# 12. OVERALL MODEL PERFORMANCE
# =============================================================================

overall_model_performance = (

    valid_results_df

    .groupby("Model")

    .agg({

        "Accuracy": "mean",

        "Balanced Accuracy": "mean",

        "Precision (Macro)": "mean",

        "Precision (Weighted)": "mean",

        "Recall (Macro)": "mean",

        "Recall (Weighted)": "mean",

        "F1 (Macro)": "mean",

        "F1 (Weighted)": "mean",

        "MCC": "mean",

        "Normalized MCC": "mean",

        "Kappa": "mean",

        "ROC AUC": "mean",

        "Composite Score": "mean",

        "Training Time": "mean",

        "Prediction Time": "mean"

    })

)


overall_model_performance = (

    overall_model_performance
    .replace([np.inf, -np.inf], np.nan)
    .round(4)

)


# =============================================================================
# 13. OVERALL COMPOSITE SCORE RANKING
# =============================================================================

overall_composite_ranking = (

    overall_model_performance

    .sort_values(
        by="Composite Score",
        ascending=False,
        na_position="last"
    )

    .reset_index()

)


# Ensure Model is a string

overall_composite_ranking["Model"] = (

    overall_composite_ranking["Model"]
    .astype(str)
    .str.strip()

)


overall_composite_ranking.insert(

    0,

    "Overall Composite Rank",

    range(
        1,
        len(overall_composite_ranking) + 1
    )

)


print("\n")
print("=" * 90)
print("OVERALL COMPOSITE SCORE RANKING")
print("=" * 90)

display(overall_composite_ranking)


# =============================================================================
# 14. AVERAGE RANK ANALYSIS
# =============================================================================
#
# Four metrics:
#
# 1. Balanced Accuracy
# 2. F1 Weighted
# 3. Normalized MCC
# 4. ROC AUC
#
# Lower Average Rank = Better
# =============================================================================

ranking_table = (

    overall_model_performance
    .copy()
    .reset_index()

)


ranking_table["Model"] = (

    ranking_table["Model"]
    .astype(str)
    .str.strip()

)


# Individual metric ranks

ranking_table["Balanced Accuracy Rank"] = (

    ranking_table["Balanced Accuracy"]
    .rank(
        ascending=False,
        method="average",
        na_option="keep"
    )

)


ranking_table["F1 Rank"] = (

    ranking_table["F1 (Weighted)"]
    .rank(
        ascending=False,
        method="average",
        na_option="keep"
    )

)


ranking_table["MCC Rank"] = (

    ranking_table["Normalized MCC"]
    .rank(
        ascending=False,
        method="average",
        na_option="keep"
    )

)


ranking_table["ROC AUC Rank"] = (

    ranking_table["ROC AUC"]
    .rank(
        ascending=False,
        method="average",
        na_option="keep"
    )

)


# Average rank

ranking_table["Average Rank"] = (

    ranking_table[
        [
            "Balanced Accuracy Rank",
            "F1 Rank",
            "MCC Rank",
            "ROC AUC Rank"
        ]
    ]

    .mean(
        axis=1,
        skipna=True
    )

)


average_rank_ranking = (

    ranking_table

    .sort_values(
        by="Average Rank",
        ascending=True,
        na_position="last"
    )

    .reset_index(drop=True)

)


average_rank_ranking.insert(

    0,

    "Overall Average-Rank Position",

    range(
        1,
        len(average_rank_ranking) + 1
    )

)


average_rank_ranking = (
    average_rank_ranking.round(4)
)


print("\n")
print("=" * 90)
print("OVERALL AVERAGE-RANK ANALYSIS")
print("=" * 90)

display(average_rank_ranking)


# =============================================================================
# 15. TOP-5 OVERALL BY COMPOSITE SCORE
# =============================================================================

top5_overall_composite = (

    overall_composite_ranking
    .head(5)
    .copy()

)


print("\n")
print("=" * 90)
print("TOP-5 OVERALL MODELS - COMPOSITE SCORE")
print("=" * 90)

display(top5_overall_composite)


# =============================================================================
# 16. TOP-5 OVERALL BY AVERAGE RANK
# =============================================================================

top5_overall_average_rank = (

    average_rank_ranking
    .head(5)
    .copy()

)


print("\n")
print("=" * 90)
print("TOP-5 OVERALL MODELS - AVERAGE RANK")
print("=" * 90)

display(top5_overall_average_rank)


# =============================================================================
# 17. TOP-5 OVERLAP ANALYSIS
# =============================================================================

top5_models_by_dataset = {}


for dataset_name, group in (

    valid_results_df

    .sort_values(
        by="Composite Score",
        ascending=False
    )

    .groupby("Dataset")

):

    top5_models_by_dataset[dataset_name] = (

        group
        .head(5)
        ["Model"]
        .astype(str)
        .str.strip()
        .tolist()

    )


# =============================================================================
# 18. OVERLAP TABLE
# =============================================================================

dataset_names = list(
    top5_models_by_dataset.keys()
)


model_names = (

    valid_results_df["Model"]
    .astype(str)
    .str.strip()
    .unique()
    .tolist()

)


overlap_records = []


for model_name in model_names:

    datasets_in_top5 = []

    positions = []


    for dataset_name in dataset_names:

        models = top5_models_by_dataset[
            dataset_name
        ]


        if model_name in models:

            datasets_in_top5.append(
                dataset_name
            )

            positions.append(

                models.index(
                    model_name
                ) + 1

            )


    overlap_count = len(
        datasets_in_top5
    )


    coverage = (

        overlap_count
        / len(dataset_names)
        * 100

    )


    mean_position = (

        np.mean(positions)
        if positions
        else np.nan

    )


    overlap_records.append({

        "Model":
            model_name,

        "Top-5 Appearances":
            overlap_count,

        "Top-5 Coverage (%)":
            coverage,

        "Mean Top-5 Position":
            mean_position,

        "Datasets in Top-5":
            ", ".join(
                datasets_in_top5
            )

    })


top5_overlap_df = pd.DataFrame(
    overlap_records
)


# =============================================================================
# 19. MERGE PERFORMANCE INFORMATION
# =============================================================================
#
# IMPORTANT:
# Model is explicitly kept as a STRING in every table.
# This prevents:
#
# ValueError:
# "merge on object and int64 columns for key 'Model'"
#
# =============================================================================

performance_for_merge = (

    overall_model_performance
    .reset_index()

)


performance_for_merge["Model"] = (

    performance_for_merge["Model"]
    .astype(str)
    .str.strip()

)


top5_overlap_df["Model"] = (

    top5_overlap_df["Model"]
    .astype(str)
    .str.strip()

)


top5_overlap_analysis = (

    top5_overlap_df

    .merge(

        performance_for_merge[
            [
                "Model",
                "Composite Score",
                "Balanced Accuracy",
                "F1 (Weighted)",
                "Normalized MCC",
                "ROC AUC"
            ]
        ],

        on="Model",

        how="left"

    )

)


# =============================================================================
# 20. MERGE AVERAGE RANK
# =============================================================================

average_rank_for_merge = (

    average_rank_ranking[
        [
            "Model",
            "Average Rank"
        ]
    ]

    .copy()

)


average_rank_for_merge["Model"] = (

    average_rank_for_merge["Model"]
    .astype(str)
    .str.strip()

)


top5_overlap_analysis = (

    top5_overlap_analysis

    .merge(

        average_rank_for_merge,

        on="Model",

        how="left"

    )

)


# =============================================================================
# 21. SORT OVERLAP ANALYSIS
# =============================================================================

top5_overlap_analysis = (

    top5_overlap_analysis

    .sort_values(

        by=[
            "Top-5 Appearances",
            "Mean Top-5 Position",
            "Composite Score"
        ],

        ascending=[
            False,
            True,
            False
        ],

        na_position="last"

    )

    .reset_index(drop=True)

)


top5_overlap_analysis.insert(

    0,

    "Overlap Rank",

    range(
        1,
        len(top5_overlap_analysis) + 1
    )

)


top5_overlap_analysis = (

    top5_overlap_analysis.round(4)

)


print("\n")
print("=" * 90)
print("TOP-5 OVERLAP / CONSISTENCY ANALYSIS")
print("=" * 90)

display(top5_overlap_analysis)


# =============================================================================
# 22. CONSENSUS SCORE
# =============================================================================
#
# Final consensus:
#
# 40% Top-5 Coverage
# 40% Composite Performance
# 20% Rank Consistency
#
# =============================================================================

consensus_df = (

    top5_overlap_analysis
    .copy()

)


# -----------------------------------------------------------------------------
# 22.1 Coverage Score
# -----------------------------------------------------------------------------

consensus_df["Coverage Score"] = (

    consensus_df["Top-5 Appearances"]
    /
    len(dataset_names)

)


# -----------------------------------------------------------------------------
# 22.2 Composite Performance Score
#
# Min-Max normalization
# -----------------------------------------------------------------------------

composite_min = (

    consensus_df["Composite Score"]
    .min()

)

composite_max = (

    consensus_df["Composite Score"]
    .max()

)


if (

    pd.notna(composite_min)

    and

    pd.notna(composite_max)

    and

    composite_max > composite_min

):

    consensus_df["Composite Performance Score"] = (

        (
            consensus_df["Composite Score"]
            - composite_min
        )
        /
        (
            composite_max
            - composite_min
        )

    )

else:

    consensus_df["Composite Performance Score"] = np.nan


# -----------------------------------------------------------------------------
# 22.3 Rank Consistency Score
#
# Lower Average Rank = Better
# -----------------------------------------------------------------------------

rank_min = (

    consensus_df["Average Rank"]
    .min()

)

rank_max = (

    consensus_df["Average Rank"]
    .max()

)


if (

    pd.notna(rank_min)

    and

    pd.notna(rank_max)

    and

    rank_max > rank_min

):

    consensus_df["Rank Consistency Score"] = (

        (
            rank_max
            - consensus_df["Average Rank"]
        )
        /
        (
            rank_max
            - rank_min
        )

    )

else:

    consensus_df["Rank Consistency Score"] = np.nan


# =============================================================================
# 23. FINAL CONSENSUS SCORE
# =============================================================================

consensus_df["Consensus Score"] = (

      0.40
      * consensus_df["Coverage Score"]

    + 0.40
      * consensus_df["Composite Performance Score"]

    + 0.20
      * consensus_df["Rank Consistency Score"]

)


consensus_df["Consensus Score"] = (

    consensus_df["Consensus Score"]
    .replace([np.inf, -np.inf], np.nan)
    .round(4)

)


# =============================================================================
# 24. FINAL CONSENSUS RANKING
# =============================================================================

final_consensus_ranking = (

    consensus_df

    .sort_values(

        by=[
            "Consensus Score",
            "Top-5 Appearances",
            "Mean Top-5 Position",
            "Composite Score",
            "Average Rank"
        ],

        ascending=[
            False,
            False,
            True,
            False,
            True
        ],

        na_position="last"

    )

    .reset_index(drop=True)

)


final_consensus_ranking.insert(

    0,

    "Final Consensus Rank",

    range(
        1,
        len(final_consensus_ranking) + 1
    )

)


print("\n")
print("=" * 90)
print("FINAL CONSENSUS RANKING")
print("=" * 90)

display(
    final_consensus_ranking.round(4)
)


# =============================================================================
# 25. FINAL OVERALL WINNER
# =============================================================================

valid_consensus = (

    final_consensus_ranking[
        final_consensus_ranking["Consensus Score"].notna()
    ]

)


if len(valid_consensus) > 0:

    overall_winner = (
        valid_consensus.iloc[0]
    )


    print("\n")
    print("=" * 90)
    print("FINAL OVERALL WINNER")
    print("=" * 90)

    print(
        "\nModel:",
        overall_winner["Model"]
    )

    print(
        "Top-5 Appearances:",
        int(overall_winner["Top-5 Appearances"])
    )

    print(
        "Top-5 Coverage:",
        f"{overall_winner['Top-5 Coverage (%)']:.2f}%"
    )

    print(
        "Mean Composite Score:",
        f"{overall_winner['Composite Score']:.4f}"
    )

    print(
        "Average Rank:",
        f"{overall_winner['Average Rank']:.4f}"
    )

    print(
        "Consensus Score:",
        f"{overall_winner['Consensus Score']:.4f}"
    )

else:

    overall_winner = None

    print(
        "\nNo valid overall winner could be identified."
    )


# =============================================================================
# 26. DATASET-LEVEL PERFORMANCE
# =============================================================================

dataset_ranking = (

    valid_results_df

    .groupby("Dataset")

    .agg({

        "Accuracy": "mean",

        "Balanced Accuracy": "mean",

        "F1 (Weighted)": "mean",

        "MCC": "mean",

        "Normalized MCC": "mean",

        "ROC AUC": "mean",

        "Composite Score": "mean"

    })

)


# =============================================================================
# 27. DATASET METRIC RANKS
# =============================================================================

dataset_ranking["Balanced Accuracy Rank"] = (

    dataset_ranking["Balanced Accuracy"]
    .rank(
        ascending=False,
        method="average",
        na_option="keep"
    )

)


dataset_ranking["F1 Rank"] = (

    dataset_ranking["F1 (Weighted)"]
    .rank(
        ascending=False,
        method="average",
        na_option="keep"
    )

)


dataset_ranking["MCC Rank"] = (

    dataset_ranking["Normalized MCC"]
    .rank(
        ascending=False,
        method="average",
        na_option="keep"
    )

)


dataset_ranking["ROC AUC Rank"] = (

    dataset_ranking["ROC AUC"]
    .rank(
        ascending=False,
        method="average",
        na_option="keep"
    )

)


dataset_ranking["Average Rank"] = (

    dataset_ranking[
        [
            "Balanced Accuracy Rank",
            "F1 Rank",
            "MCC Rank",
            "ROC AUC Rank"
        ]
    ]

    .mean(
        axis=1,
        skipna=True
    )

)


# =============================================================================
# 28. FINAL DATASET RANKING
# =============================================================================

dataset_ranking = (

    dataset_ranking

    .sort_values(

        by=[
            "Composite Score",
            "Average Rank"
        ],

        ascending=[
            False,
            True
        ],

        na_position="last"

    )

    .reset_index()

)


dataset_ranking.insert(

    0,

    "Dataset Composite Rank",

    range(
        1,
        len(dataset_ranking) + 1
    )

)


dataset_ranking = (
    dataset_ranking.round(4)
)


print("\n")
print("=" * 90)
print("DATASET RANKING")
print("=" * 90)

display(dataset_ranking)


# =============================================================================
# 29. BEST MODEL FOR EACH DATASET
# =============================================================================

best_model_per_dataset = (

    valid_results_df

    .sort_values(
        by="Composite Score",
        ascending=False
    )

    .groupby(
        "Dataset",
        group_keys=False
    )

    .head(1)

    .reset_index(drop=True)

)


print("\n")
print("=" * 90)
print("BEST MODEL FOR EACH DATASET")
print("=" * 90)

display(best_model_per_dataset)


# =============================================================================
# 30. BEST DATASET FOR EACH MODEL
# =============================================================================

best_dataset_per_model = (

    valid_results_df

    .sort_values(
        by="Composite Score",
        ascending=False
    )

    .groupby(
        "Model",
        group_keys=False
    )

    .head(1)

    .reset_index(drop=True)

)


print("\n")
print("=" * 90)
print("BEST DATASET FOR EACH MODEL")
print("=" * 90)

display(best_dataset_per_model)


# =============================================================================
# 31. DATASET SUMMARY
# =============================================================================

dataset_summary_records = []


for dataset_name, dataset in prepared_datasets.items():

    dataset_summary_records.append({

        "Dataset":
            dataset_name,

        "Samples":
            dataset["X"].shape[0],

        "Training Samples":
            dataset["X_train"].shape[0],

        "Testing Samples":
            dataset["X_test"].shape[0],

        "Features":
            dataset["X"].shape[1],

        "Classes":
            len(
                dataset["target_encoder"].classes_
            )

    })


dataset_summary = pd.DataFrame(
    dataset_summary_records
)


# =============================================================================
# 32. SAVE EXCEL WORKBOOK
# =============================================================================

excel_file = os.path.join(

    results_folder,

    "Benchmark_Results.xlsx"

)


with pd.ExcelWriter(

    excel_file,

    engine="openpyxl"

) as writer:

    all_results_df.to_excel(
        writer,
        sheet_name="All Results",
        index=False
    )

    classwise_results_df.to_excel(
        writer,
        sheet_name="Classwise Results",
        index=False
    )

    top5_per_dataset.to_excel(
        writer,
        sheet_name="Top5 Per Dataset",
        index=False
    )

    overall_composite_ranking.to_excel(
        writer,
        sheet_name="Composite Ranking",
        index=False
    )

    average_rank_ranking.to_excel(
        writer,
        sheet_name="Average Rank",
        index=False
    )

    top5_overlap_analysis.to_excel(
        writer,
        sheet_name="Top5 Overlap",
        index=False
    )

    final_consensus_ranking.to_excel(
        writer,
        sheet_name="Final Consensus",
        index=False
    )

    dataset_ranking.to_excel(
        writer,
        sheet_name="Dataset Ranking",
        index=False
    )

    best_model_per_dataset.to_excel(
        writer,
        sheet_name="Best Model Dataset",
        index=False
    )

    best_dataset_per_model.to_excel(
        writer,
        sheet_name="Best Dataset Model",
        index=False
    )

    dataset_summary.to_excel(
        writer,
        sheet_name="Dataset Summary",
        index=False
    )


# =============================================================================
# 33. SAVE INDIVIDUAL CSV FILES
# =============================================================================

csv_outputs = {

    "01_All_Results.csv":
        all_results_df,

    "02_Classwise_Results.csv":
        classwise_results_df,

    "03_Top5_Per_Dataset.csv":
        top5_per_dataset,

    "04_Overall_Composite_Ranking.csv":
        overall_composite_ranking,

    "05_Overall_Average_Rank.csv":
        average_rank_ranking,

    "06_Top5_Overlap_Analysis.csv":
        top5_overlap_analysis,

    "07_Final_Consensus_Ranking.csv":
        final_consensus_ranking,

    "08_Dataset_Ranking.csv":
        dataset_ranking,

    "09_Best_Model_Per_Dataset.csv":
        best_model_per_dataset,

    "10_Best_Dataset_Per_Model.csv":
        best_dataset_per_model,

    "11_Dataset_Summary.csv":
        dataset_summary

}


for filename, dataframe in csv_outputs.items():

    dataframe.to_csv(

        os.path.join(
            results_folder,
            filename
        ),

        index=False

    )


# =============================================================================
# 34. SAVE FINAL WINNER
# =============================================================================

if overall_winner is not None:

    winner_summary = pd.DataFrame({

        "Overall Winner": [
            overall_winner["Model"]
        ],

        "Top-5 Appearances": [
            overall_winner["Top-5 Appearances"]
        ],

        "Top-5 Coverage (%)": [
            overall_winner["Top-5 Coverage (%)"]
        ],

        "Mean Composite Score": [
            overall_winner["Composite Score"]
        ],

        "Average Rank": [
            overall_winner["Average Rank"]
        ],

        "Consensus Score": [
            overall_winner["Consensus Score"]
        ]

    })

else:

    winner_summary = pd.DataFrame({

        "Overall Winner": [
            "No valid winner"
        ]

    })


winner_summary.to_csv(

    os.path.join(
        results_folder,
        "12_Overall_Winner.csv"
    ),

    index=False

)


# =============================================================================
# 35. FINAL SUMMARY
# =============================================================================

print("\n")
print("=" * 90)
print("MODULE 8 COMPLETED SUCCESSFULLY")
print("=" * 90)

print(
    "\nNumber of datasets:",
    overall_results_df["Dataset"].nunique()
)

print(
    "Number of classifiers:",
    overall_results_df["Model"].nunique()
)

print(
    "Total experiments:",
    len(overall_results_df)
)

print(
    "Valid experiments:",
    overall_results_df["Valid Composite"].sum()
)

print(
    "Invalid experiments:",
    (~overall_results_df["Valid Composite"]).sum()
)

if overall_winner is not None:

    print(
        "\nFINAL OVERALL WINNER:",
        overall_winner["Model"]
    )

print(
    "\nExcel workbook:"
)

print(
    excel_file
)

print(
    "\nAll results saved in:"
)

print(
    results_folder
)

print("=" * 90)
```

```python
# =============================================================================
# MODULE 9
# FEATURE CATEGORY ABLATION ANALYSIS
# =============================================================================
#
# OBJECTIVE:
# Examine the effect of progressively adding feature categories on
# predictive performance using the five consensus classifiers.
#
# DATASETS:
#   1. 3class_original
#   2. 3class_duplicate_removed
#   3. 2class_original
#   4. 2class_duplicate_removed
#
# CLASSIFIERS:
#   1. Gradient Boosting
#   2. Random Forest
#   3. Quadratic Discriminant Analysis
#   4. Extra Trees
#   5. Decision Tree
#
# COMPOSITE SCORE:
#
# CS = 0.30(Balanced Accuracy)
#    + 0.30(F1 Weighted)
#    + 0.20((MCC + 1) / 2)
#    + 0.20(ROC-AUC)
#
# PRIMARY ANALYSIS:
# Mean performance across the FIVE classifiers for each
# dataset + feature configuration.
#
# IMPORTANT:
# The existing Module 4 train/test split is reused.
# No new train/test split is created here.
# =============================================================================


import os
import time
import copy
import numpy as np
import pandas as pd

from datetime import datetime

from sklearn.ensemble import (
    GradientBoostingClassifier,
    RandomForestClassifier,
    ExtraTreesClassifier
)

from sklearn.tree import DecisionTreeClassifier

from sklearn.discriminant_analysis import (
    QuadraticDiscriminantAnalysis
)

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    cohen_kappa_score,
    roc_auc_score,
    confusion_matrix
)

from sklearn.preprocessing import label_binarize


print("=" * 80)
print("MODULE 9 : FEATURE CATEGORY ABLATION ANALYSIS")
print("=" * 80)


# =============================================================================
# 1. TARGET VARIABLE
# =============================================================================

TARGET_COLUMN = target_column

print("\nTarget variable:", TARGET_COLUMN)


# =============================================================================
# 2. DEFINE FEATURE CATEGORIES
# =============================================================================
#
# NOTE:
# CD_Div is the target variable and therefore is NOT included here.
#
# Based on the feature structure already established in the dataset:
#
# Demographic:
#   Gender
#
# Academic:
#   Discipline_id
#   Reg_Private
#   PD_Div
#   PD_Att
#   CD_ufm_case
#   PD_ufm_case
#   PD_COMBI_name
#
# Socio-economic:
#   CD_INSTIT_TYPE
#   CD_DIST
#   PD_INSTIT_TYPE
#   Poverty
#   HDI
#   Literacy_Rate
#   PD_DIST
#
# =============================================================================


DEMOGRAPHIC_FEATURES = [
    "Gender"
]


ACADEMIC_FEATURES = [
    "Discipline_id",
    "Reg_Private",
    "PD_Div",
    "PD_Att",
    "CD_ufm_case",
    "PD_ufm_case",
    "PD_COMBI_name"
]


SOCIOECONOMIC_FEATURES = [
    "CD_INSTIT_TYPE",
    "CD_DIST",
    "PD_INSTIT_TYPE",
    "Poverty",
    "HDI",
    "Literacy_Rate",
    "PD_DIST"
]


# =============================================================================
# 3. CHECK FEATURE DEFINITIONS
# =============================================================================

all_ablation_features = list(
    dict.fromkeys(
        ACADEMIC_FEATURES
        + DEMOGRAPHIC_FEATURES
        + SOCIOECONOMIC_FEATURES
    )
)


if TARGET_COLUMN in all_ablation_features:

    raise ValueError(
        f"ERROR: Target variable '{TARGET_COLUMN}' "
        "has been included in the predictor features."
    )


print("\nFeature category definitions verified.")


# =============================================================================
# 4. DEFINE SIX FEATURE CONFIGURATIONS
# =============================================================================
#
# FA1 = Academic
# FA2 = Academic + Demographic
# FA3 = Academic + Socio-economic
# FA4 = Academic + Demographic + Socio-economic
# FA5 = Demographic
# FA6 = Socio-economic
#
# FA1 -> FA2 -> FA4
# tests incremental contribution of Demographic features.
#
# FA1 -> FA3 -> FA4
# tests incremental contribution of Socio-economic features.
#
# FA5 and FA6 are control configurations.
# =============================================================================


feature_configurations = {

    "FA1_Academic":
        ACADEMIC_FEATURES,

    "FA2_Academic_Demographic":
        ACADEMIC_FEATURES
        + DEMOGRAPHIC_FEATURES,

    "FA3_Academic_Socioeconomic":
        ACADEMIC_FEATURES
        + SOCIOECONOMIC_FEATURES,

    "FA4_Academic_Demographic_Socioeconomic":
        ACADEMIC_FEATURES
        + DEMOGRAPHIC_FEATURES
        + SOCIOECONOMIC_FEATURES,

    "FA5_Demographic":
        DEMOGRAPHIC_FEATURES,

    "FA6_Socioeconomic":
        SOCIOECONOMIC_FEATURES
}


# =============================================================================
# 5. DISPLAY FEATURE CONFIGURATIONS
# =============================================================================

print("\n" + "=" * 80)
print("FEATURE CONFIGURATIONS")
print("=" * 80)

for name, features in feature_configurations.items():

    print(f"\n{name}")
    print(f"Number of features: {len(features)}")
    print("Features:", ", ".join(features))


# =============================================================================
# 6. DATASETS FOR FEATURE ABLATION
# =============================================================================
#
# Only ORIGINAL and DUPLICATE-REMOVED datasets are used.
#
# Outlier-removed datasets are deliberately excluded from this experiment.
# =============================================================================


ablation_datasets = [
    "3class_original",
    "3class_duplicate_removed",
    "2class_original",
    "2class_duplicate_removed"
]


# =============================================================================
# 7. VERIFY DATASETS
# =============================================================================

for dataset_name in ablation_datasets:

    if dataset_name not in prepared_datasets:

        raise ValueError(
            f"Dataset '{dataset_name}' was not found "
            "in prepared_datasets."
        )


print("\nAll required datasets are available.")


# =============================================================================
# 8. VERIFY FEATURES AGAINST DATA
# =============================================================================

for dataset_name in ablation_datasets:

    X_available = prepared_datasets[
        dataset_name
    ]["X"]

    missing_features = [
        feature
        for feature in all_ablation_features
        if feature not in X_available.columns
    ]

    if missing_features:

        raise ValueError(
            f"\nMissing features in {dataset_name}:\n"
            + "\n".join(missing_features)
        )


print("All required features are available.")


# =============================================================================
# 9. DEFINE FIVE CONSENSUS CLASSIFIERS
# =============================================================================

ablation_models = {

    "Gradient Boosting":
        GradientBoostingClassifier(
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            n_jobs=-1
        ),

    "Quadratic Discriminant Analysis":
        QuadraticDiscriminantAnalysis(),

    "Extra Trees":
        ExtraTreesClassifier(
            n_estimators=300,
            random_state=42,
            n_jobs=-1
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            random_state=42
        )
}


print("\n" + "=" * 80)
print("CONSENSUS CLASSIFIERS")
print("=" * 80)

for i, model_name in enumerate(
    ablation_models.keys(),
    start=1
):
    print(f"{i}. {model_name}")


# =============================================================================
# 10. RESULT STORAGE
# =============================================================================

ablation_results = []

classwise_results = []


# =============================================================================
# 11. TOTAL EXPERIMENTS
# =============================================================================

total_experiments = (
    len(ablation_datasets)
    * len(feature_configurations)
    * len(ablation_models)
)

experiment_number = 0

print(
    f"\nTotal experiments: {total_experiments}"
)

print(
    "4 datasets × 6 feature configurations × "
    "5 classifiers = 120 experiments"
)


# =============================================================================
# 12. MAIN FEATURE ABLATION EXPERIMENT
# =============================================================================

for dataset_name in ablation_datasets:

    print("\n" + "=" * 80)
    print(f"DATASET: {dataset_name}")
    print("=" * 80)


    dataset = prepared_datasets[
        dataset_name
    ]


    # -------------------------------------------------------------------------
    # IMPORTANT:
    # Reuse the exact train/test partitions produced by Module 4.
    # -------------------------------------------------------------------------

    X_train_full = dataset["X_train"]
    X_test_full = dataset["X_test"]

    y_train = dataset["y_train"]
    y_test = dataset["y_test"]

    target_encoder = dataset["target_encoder"]


    for configuration_name, selected_features in (
        feature_configurations.items()
    ):


        # ---------------------------------------------------------------------
        # Select features from the EXISTING train/test datasets.
        # ---------------------------------------------------------------------

        X_train = X_train_full[
            selected_features
        ].copy()

        X_test = X_test_full[
            selected_features
        ].copy()


        # ---------------------------------------------------------------------
        # Verify target is not present
        # ---------------------------------------------------------------------

        if TARGET_COLUMN in X_train.columns:

            raise ValueError(
                f"Target leakage detected in {configuration_name}"
            )


        print("\n" + "-" * 80)
        print(
            f"Configuration: {configuration_name}"
        )


        for model_name, base_model in (
            ablation_models.items()
        ):

            experiment_number += 1

            print(
                f"\n[{experiment_number}/{total_experiments}] "
                f"{model_name}"
            )


            try:

                # -----------------------------------------------------------------
                # Fresh model instance
                # -----------------------------------------------------------------

                model = copy.deepcopy(
                    base_model
                )


                # -----------------------------------------------------------------
                # TRAIN
                # -----------------------------------------------------------------

                start_time = time.time()

                model.fit(
                    X_train,
                    y_train
                )

                training_time = (
                    time.time() - start_time
                )


                # -----------------------------------------------------------------
                # PREDICT
                # -----------------------------------------------------------------

                start_time = time.time()

                y_pred = model.predict(
                    X_test
                )

                prediction_time = (
                    time.time() - start_time
                )


                # -----------------------------------------------------------------
                # SCORE / PROBABILITY
                # -----------------------------------------------------------------

                y_score = None

                if hasattr(
                    model,
                    "predict_proba"
                ):

                    y_score = model.predict_proba(
                        X_test
                    )

                elif hasattr(
                    model,
                    "decision_function"
                ):

                    y_score = model.decision_function(
                        X_test
                    )


                # -----------------------------------------------------------------
                # METRICS
                # -----------------------------------------------------------------

                accuracy = accuracy_score(
                    y_test,
                    y_pred
                )


                balanced_accuracy = (
                    balanced_accuracy_score(
                        y_test,
                        y_pred
                    )
                )


                precision_macro = precision_score(
                    y_test,
                    y_pred,
                    average="macro",
                    zero_division=0
                )


                precision_weighted = precision_score(
                    y_test,
                    y_pred,
                    average="weighted",
                    zero_division=0
                )


                recall_macro = recall_score(
                    y_test,
                    y_pred,
                    average="macro",
                    zero_division=0
                )


                recall_weighted = recall_score(
                    y_test,
                    y_pred,
                    average="weighted",
                    zero_division=0
                )


                f1_macro = f1_score(
                    y_test,
                    y_pred,
                    average="macro",
                    zero_division=0
                )


                f1_weighted = f1_score(
                    y_test,
                    y_pred,
                    average="weighted",
                    zero_division=0
                )


                mcc = matthews_corrcoef(
                    y_test,
                    y_pred
                )


                kappa = cohen_kappa_score(
                    y_test,
                    y_pred
                )


                # -----------------------------------------------------------------
                # ROC-AUC
                # -----------------------------------------------------------------

                roc_auc = np.nan

                number_of_classes = len(
                    target_encoder.classes_
                )


                try:

                    if y_score is not None:

                        if number_of_classes == 2:

                            if np.ndim(y_score) == 1:

                                roc_auc = roc_auc_score(
                                    y_test,
                                    y_score
                                )

                            else:

                                roc_auc = roc_auc_score(
                                    y_test,
                                    y_score[:, 1]
                                )

                        else:

                            y_test_binary = label_binarize(
                                y_test,
                                classes=np.arange(
                                    number_of_classes
                                )
                            )

                            roc_auc = roc_auc_score(
                                y_test_binary,
                                y_score,
                                average="weighted",
                                multi_class="ovr"
                            )

                except Exception:

                    roc_auc = np.nan


                # -----------------------------------------------------------------
                # NORMALIZED MCC
                # -----------------------------------------------------------------

                normalized_mcc = (
                    (mcc + 1) / 2
                )


                # -----------------------------------------------------------------
                # COMPOSITE SCORE
                # -----------------------------------------------------------------

                if pd.notna(roc_auc):

                    composite_score = (

                        0.30
                        * balanced_accuracy

                        +

                        0.30
                        * f1_weighted

                        +

                        0.20
                        * normalized_mcc

                        +

                        0.20
                        * roc_auc

                    )

                else:

                    composite_score = np.nan


                # -----------------------------------------------------------------
                # SAVE OVERALL RESULTS
                # -----------------------------------------------------------------

                ablation_results.append({

                    "Dataset":
                        dataset_name,

                    "Feature Configuration":
                        configuration_name,

                    "Feature Count":
                        len(selected_features),

                    "Model":
                        model_name,

                    "Accuracy":
                        accuracy,

                    "Balanced Accuracy":
                        balanced_accuracy,

                    "Precision (Macro)":
                        precision_macro,

                    "Precision (Weighted)":
                        precision_weighted,

                    "Recall (Macro)":
                        recall_macro,

                    "Recall (Weighted)":
                        recall_weighted,

                    "F1 (Macro)":
                        f1_macro,

                    "F1 (Weighted)":
                        f1_weighted,

                    "MCC":
                        mcc,

                    "Kappa":
                        kappa,

                    "ROC AUC":
                        roc_auc,

                    "Composite Score":
                        composite_score,

                    "Training Time":
                        training_time,

                    "Prediction Time":
                        prediction_time
                })


                # -----------------------------------------------------------------
                # CLASSWISE RESULTS
                # -----------------------------------------------------------------

                cm = confusion_matrix(
                    y_test,
                    y_pred
                )

                class_names = (
                    target_encoder.classes_
                )


                for i, class_name in enumerate(
                    class_names
                ):

                    TP = cm[i, i]

                    FP = (
                        cm[:, i].sum()
                        - TP
                    )

                    FN = (
                        cm[i, :].sum()
                        - TP
                    )

                    TN = (
                        cm.sum()
                        - TP
                        - FP
                        - FN
                    )


                    precision_class = (
                        TP / (TP + FP)
                        if (TP + FP) > 0
                        else 0
                    )


                    sensitivity = (
                        TP / (TP + FN)
                        if (TP + FN) > 0
                        else 0
                    )


                    specificity = (
                        TN / (TN + FP)
                        if (TN + FP) > 0
                        else 0
                    )


                    f1_class = (

                        2
                        * precision_class
                        * sensitivity
                        /
                        (
                            precision_class
                            + sensitivity
                        )

                        if (
                            precision_class
                            + sensitivity
                        ) > 0

                        else 0
                    )


                    classwise_results.append({

                        "Dataset":
                            dataset_name,

                        "Feature Configuration":
                            configuration_name,

                        "Model":
                            model_name,

                        "Class":
                            class_name,

                        "TP":
                            TP,

                        "TN":
                            TN,

                        "FP":
                            FP,

                        "FN":
                            FN,

                        "Precision":
                            precision_class,

                        "Sensitivity":
                            sensitivity,

                        "Specificity":
                            specificity,

                        "F1":
                            f1_class
                    })


                # -----------------------------------------------------------------
                # DISPLAY MAIN METRICS
                # -----------------------------------------------------------------

                print(
                    f"Accuracy        : {accuracy:.4f}"
                )

                print(
                    f"Balanced Acc.  : "
                    f"{balanced_accuracy:.4f}"
                )

                print(
                    f"Weighted F1    : "
                    f"{f1_weighted:.4f}"
                )

                print(
                    f"MCC            : "
                    f"{mcc:.4f}"
                )

                if pd.notna(roc_auc):

                    print(
                        f"ROC-AUC        : "
                        f"{roc_auc:.4f}"
                    )

                else:

                    print(
                        "ROC-AUC        : NA"
                    )


                if pd.notna(composite_score):

                    print(
                        f"Composite Score: "
                        f"{composite_score:.4f}"
                    )

                else:

                    print(
                        "Composite Score: NA"
                    )


            except Exception as error:

                print(
                    f"\nERROR in {model_name}"
                )

                print(
                    f"Dataset       : {dataset_name}"
                )

                print(
                    f"Configuration : {configuration_name}"
                )

                print(
                    f"Error         : {error}"
                )


# =============================================================================
# 13. CREATE RESULT DATAFRAMES
# =============================================================================

ablation_results_df = pd.DataFrame(
    ablation_results
)

classwise_results_df = pd.DataFrame(
    classwise_results
)


# =============================================================================
# 14. PRIMARY FEATURE CONFIGURATION SUMMARY
# =============================================================================
#
# IMPORTANT:
# The mean is calculated across the FIVE CLASSIFIERS.
#
# Therefore:
#
# FA1 Mean Composite Score =
#
#   (GB + RF + QDA + ET + DT) / 5
#
# =============================================================================

configuration_summary = (

    ablation_results_df

    .groupby(
        [
            "Dataset",
            "Feature Configuration"
        ]
    )

    .agg(

        Classifiers=(
            "Model",
            "nunique"
        ),

        Mean_Accuracy=(
            "Accuracy",
            "mean"
        ),

        SD_Accuracy=(
            "Accuracy",
            "std"
        ),

        Mean_Balanced_Accuracy=(
            "Balanced Accuracy",
            "mean"
        ),

        SD_Balanced_Accuracy=(
            "Balanced Accuracy",
            "std"
        ),

        Mean_F1_Weighted=(
            "F1 (Weighted)",
            "mean"
        ),

        SD_F1_Weighted=(
            "F1 (Weighted)",
            "std"
        ),

        Mean_MCC=(
            "MCC",
            "mean"
        ),

        SD_MCC=(
            "MCC",
            "std"
        ),

        Mean_ROC_AUC=(
            "ROC AUC",
            "mean"
        ),

        SD_ROC_AUC=(
            "ROC AUC",
            "std"
        ),

        Mean_Composite_Score=(
            "Composite Score",
            "mean"
        ),

        SD_Composite_Score=(
            "Composite Score",
            "std"
        )
    )

    .reset_index()
)


# =============================================================================
# 15. RANK FEATURE CONFIGURATIONS
# =============================================================================

configuration_summary[
    "Feature_Configuration_Rank"
] = (

    configuration_summary

    .groupby("Dataset")[
        "Mean_Composite_Score"
    ]

    .rank(
        ascending=False,
        method="min"
    )
)


# =============================================================================
# 16. INCREMENTAL ANALYSIS
# =============================================================================
#
# Main comparisons:
#
# FA1 -> FA2 -> FA4
#
# FA1 -> FA3 -> FA4
#
# =============================================================================

incremental_results = []


for dataset_name in ablation_datasets:

    data = configuration_summary[
        configuration_summary["Dataset"]
        == dataset_name
    ]


    scores = data.set_index(
        "Feature Configuration"
    )["Mean_Composite_Score"]


    fa1 = scores.get(
        "FA1_Academic",
        np.nan
    )

    fa2 = scores.get(
        "FA2_Academic_Demographic",
        np.nan
    )

    fa3 = scores.get(
        "FA3_Academic_Socioeconomic",
        np.nan
    )

    fa4 = scores.get(
        "FA4_Academic_Demographic_Socioeconomic",
        np.nan
    )


    incremental_results.append({

        "Dataset":
            dataset_name,

        "FA1 Academic":
            fa1,

        "FA2 Academic + Demographic":
            fa2,

        "FA3 Academic + Socioeconomic":
            fa3,

        "FA4 Full":
            fa4,

        "Gain FA1_to_FA2":
            fa2 - fa1,

        "Gain FA1_to_FA3":
            fa3 - fa1,

        "Gain FA2_to_FA4":
            fa4 - fa2,

        "Gain FA3_to_FA4":
            fa4 - fa3,

        "Total Gain FA1_to_FA4":
            fa4 - fa1,

        "Improvement FA1_to_FA2_Percent":
            ((fa2 - fa1) / fa1 * 100)
            if fa1 != 0 else np.nan,

        "Improvement FA1_to_FA3_Percent":
            ((fa3 - fa1) / fa1 * 100)
            if fa1 != 0 else np.nan,

        "Improvement FA2_to_FA4_Percent":
            ((fa4 - fa2) / fa2 * 100)
            if fa2 != 0 else np.nan,

        "Improvement FA3_to_FA4_Percent":
            ((fa4 - fa3) / fa3 * 100)
            if fa3 != 0 else np.nan,

        "Total Improvement FA1_to_FA4_Percent":
            ((fa4 - fa1) / fa1 * 100)
            if fa1 != 0 else np.nan
    })


incremental_analysis_df = pd.DataFrame(
    incremental_results
)


# =============================================================================
# 17. FEATURE CONFIGURATION RANKING
# =============================================================================

feature_configuration_ranking = (

    configuration_summary

    .sort_values(
        [
            "Dataset",
            "Mean_Composite_Score"
        ],
        ascending=[
            True,
            False
        ]
    )

    .copy()
)


feature_configuration_ranking[
    "Rank"
] = (

    feature_configuration_ranking

    .groupby("Dataset")

    .cumcount()

    + 1
)


# =============================================================================
# 18. BEST FEATURE CONFIGURATION
# =============================================================================

best_configuration = (

    feature_configuration_ranking

    .groupby(
        "Dataset",
        as_index=False
    )

    .first()
)


# =============================================================================
# 19. SECONDARY CLASSIFIER SUMMARY
# =============================================================================
#
# This is NOT the primary feature-ablation ranking.
#
# It answers:
# Which of the five classifiers is most robust across FA1–FA6?
#
# =============================================================================

classifier_summary = (

    ablation_results_df

    .groupby(
        [
            "Dataset",
            "Model"
        ]
    )

    .agg(

        Mean_Accuracy=(
            "Accuracy",
            "mean"
        ),

        Mean_Balanced_Accuracy=(
            "Balanced Accuracy",
            "mean"
        ),

        Mean_F1_Weighted=(
            "F1 (Weighted)",
            "mean"
        ),

        Mean_MCC=(
            "MCC",
            "mean"
        ),

        Mean_ROC_AUC=(
            "ROC AUC",
            "mean"
        ),

        Mean_Composite_Score=(
            "Composite Score",
            "mean"
        ),

        SD_Composite_Score=(
            "Composite Score",
            "std"
        )
    )

    .reset_index()
)


classifier_summary[
    "Classifier_Rank"
] = (

    classifier_summary

    .groupby("Dataset")[
        "Mean_Composite_Score"
    ]

    .rank(
        ascending=False,
        method="min"
    )
)


# =============================================================================
# 20. COMPOSITE SCORE MATRIX
# =============================================================================

composite_matrix = (

    configuration_summary

    .pivot(
        index="Dataset",
        columns="Feature Configuration",
        values="Mean_Composite_Score"
    )

    .reset_index()
)


# =============================================================================
# 21. DISPLAY PRIMARY RESULTS
# =============================================================================

print("\n" + "=" * 80)
print("PRIMARY FEATURE ABLATION SUMMARY")
print("=" * 80)

display(
    configuration_summary.sort_values(
        [
            "Dataset",
            "Feature_Configuration_Rank"
        ]
    )
)


print("\n" + "=" * 80)
print("INCREMENTAL FEATURE IMPROVEMENT")
print("=" * 80)

display(
    incremental_analysis_df
)


print("\n" + "=" * 80)
print("FEATURE CONFIGURATION RANKING")
print("=" * 80)

display(
    feature_configuration_ranking
)


print("\n" + "=" * 80)
print("BEST FEATURE CONFIGURATION")
print("=" * 80)

display(
    best_configuration
)


print("\n" + "=" * 80)
print("SECONDARY CLASSIFIER SUMMARY")
print("=" * 80)

display(
    classifier_summary.sort_values(
        [
            "Dataset",
            "Classifier_Rank"
        ]
    )
)


print("\n" + "=" * 80)
print("COMPOSITE SCORE MATRIX")
print("=" * 80)

display(
    composite_matrix
)


# =============================================================================
# 22. SAVE RESULTS
# =============================================================================

# -------------------------------------------------------------------------
# Find dataset directory
# -------------------------------------------------------------------------
#
# Use the path variable that already exists in the previous modules.
# If dataset_path exists, use it.
# Otherwise, if file_path exists, use it.
# Otherwise, use the current working directory.
# -------------------------------------------------------------------------


if "dataset_path" in globals():

    base_directory = os.path.dirname(
        dataset_path
    )

elif "file_path" in globals():

    base_directory = os.path.dirname(
        file_path
    )

else:

    base_directory = os.getcwd()


timestamp = datetime.now().strftime(
    "%Y%m%d_%H%M%S"
)


results_directory = os.path.join(
    base_directory,
    "Results",
    f"Feature_Ablation_{timestamp}"
)


os.makedirs(
    results_directory,
    exist_ok=True
)


print("\n" + "=" * 80)
print("SAVING FEATURE ABLATION RESULTS")
print("=" * 80)

print(
    "\nResults directory:"
)

print(
    results_directory
)


# =============================================================================
# 23. SAVE CSV RESULTS
# =============================================================================

ablation_results_df.to_csv(
    os.path.join(
        results_directory,
        "01_Detailed_Ablation_Results.csv"
    ),
    index=False
)


classwise_results_df.to_csv(
    os.path.join(
        results_directory,
        "02_Classwise_Ablation_Results.csv"
    ),
    index=False
)


configuration_summary.to_csv(
    os.path.join(
        results_directory,
        "03_Primary_Feature_Configuration_Summary.csv"
    ),
    index=False
)


incremental_analysis_df.to_csv(
    os.path.join(
        results_directory,
        "04_Incremental_Feature_Improvement.csv"
    ),
    index=False
)


feature_configuration_ranking.to_csv(
    os.path.join(
        results_directory,
        "05_Feature_Configuration_Ranking.csv"
    ),
    index=False
)


best_configuration.to_csv(
    os.path.join(
        results_directory,
        "06_Best_Feature_Configuration.csv"
    ),
    index=False
)


classifier_summary.to_csv(
    os.path.join(
        results_directory,
        "07_Classifier_Robustness_Summary.csv"
    ),
    index=False
)


composite_matrix.to_csv(
    os.path.join(
        results_directory,
        "08_Composite_Score_Matrix.csv"
    ),
    index=False
)


# =============================================================================
# 24. SAVE COMPLETE EXCEL WORKBOOK
# =============================================================================

excel_path = os.path.join(
    results_directory,
    "Feature_Ablation_Complete_Results.xlsx"
)


with pd.ExcelWriter(
    excel_path,
    engine="openpyxl"
) as writer:

    ablation_results_df.to_excel(
        writer,
        sheet_name="Detailed Results",
        index=False
    )

    classwise_results_df.to_excel(
        writer,
        sheet_name="Classwise Results",
        index=False
    )

    configuration_summary.to_excel(
        writer,
        sheet_name="Primary FA Summary",
        index=False
    )

    incremental_analysis_df.to_excel(
        writer,
        sheet_name="Incremental Analysis",
        index=False
    )

    feature_configuration_ranking.to_excel(
        writer,
        sheet_name="FA Ranking",
        index=False
    )

    best_configuration.to_excel(
        writer,
        sheet_name="Best Configuration",
        index=False
    )

    classifier_summary.to_excel(
        writer,
        sheet_name="Classifier Summary",
        index=False
    )

    composite_matrix.to_excel(
        writer,
        sheet_name="Composite Matrix",
        index=False
    )


# =============================================================================
# 25. FINAL VALIDATION
# =============================================================================

expected_experiments = (
    len(ablation_datasets)
    * len(feature_configurations)
    * len(ablation_models)
)


actual_experiments = len(
    ablation_results_df
)


print("\n" + "=" * 80)
print("EXPERIMENT VALIDATION")
print("=" * 80)

print(
    f"Expected experiments : {expected_experiments}"
)

print(
    f"Completed experiments: {actual_experiments}"
)


if actual_experiments == expected_experiments:

    print(
        "\nSUCCESS: All 120 experiments completed."
    )

else:

    print(
        "\nWARNING: The number of completed experiments "
        "does not equal 120."
    )


print("\n" + "=" * 80)
print("MODULE 9 COMPLETED")
print("=" * 80)

print(
    "\nResults saved at:"
)

print(
    results_directory
)

print(
    "\nExcel file:"
)

print(
    excel_path
)
```
